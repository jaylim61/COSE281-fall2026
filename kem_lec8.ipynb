# Lecture 8: Laplace Transform

```python
from sympy import *
```

```python
t, s, w = symbols('t, s, w', positive=True)
```

```python
Y = laplace_transform(t*cos(w*t), t, s)
simplify(Y[0])
```

```python
Y = laplace_transform(t*sin(w*t), t, s)
simplify(Y[0])
```

```python
y = inverse_laplace_transform(1 / (s-2)**3, s, t)
y
```

```python
y = inverse_laplace_transform(s / (s-2)**3, s, t); y
```

```python
simplify(y)
```

```python
simplify(simplify(y))
```
