## Laplace transform

```python
from sympy import *
```

```python
t, s = symbols('t, s', positive=True)
```

```python
laplace_transform(-4*t**2 + 16*t + 9, t, s)[0]
```

```python
laplace_transform((2*t - 1)**3, t, s)[0]
```

```python
laplace_transform(1 + exp(4*t), t, s)[0]
```

```python
laplace_transform((1 + exp(2*t))**2, t, s)[0]
```

```python
laplace_transform(t**2 - exp(-9*t) + 5, t, s)[0]
```

```python
laplace_transform((exp(t) - exp(-t))**2, t, s)[0]
```

```python
laplace_transform(cos(5*t) + sin(2*t), t, s)[0]
```

```python
laplace_transform(4*t**2 - 5*sin(3*t), t, s)[0]
```
